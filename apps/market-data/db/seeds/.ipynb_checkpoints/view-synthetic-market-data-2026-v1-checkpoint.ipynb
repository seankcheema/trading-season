# 2026-v1 synthetic market-data explorer

This notebook reads the validated Parquet archive directly. Range charts use tick-derived candles and query only aggregated results; the one-second explorer opens one daily tick partition at a time. PostgreSQL is not required.

```python

from __future__ import annotations

import json
from datetime import date, datetime, timedelta
from pathlib import Path
from zoneinfo import ZoneInfo

import duckdb
import ipywidgets as widgets
import pandas as pd
import plotly.graph_objects as go
from IPython.display import HTML, display
from plotly.subplots import make_subplots

CHICAGO = ZoneInfo('America/Chicago')
DATASET_NAME = 'synthetic-market-data-2026-v1'
candidates = [Path.cwd() / DATASET_NAME, Path.cwd() / 'apps/business-backend/db/seeds' / DATASET_NAME]
DATASET = next((path.resolve() for path in candidates if (path / 'manifest.json').is_file()), None)
if DATASET is None:
    searched = '\n'.join(f'  - {path.resolve()}' for path in candidates)
    raise FileNotFoundError(f'{DATASET_NAME}/manifest.json was not found. Generate the archive first. Searched:\n{searched}')

MANIFEST_PATH = DATASET / 'manifest.json'
manifest = json.loads(MANIFEST_PATH.read_text(encoding='utf-8'))
required = {'dataset_id', 'symbols', 'tick_files', 'candle_files', 'tick_count', 'candle_count'}
missing = sorted(required - manifest.keys())
if missing:
    raise ValueError(f'Manifest is missing required fields: {missing}')
if not manifest['tick_files'] or not manifest['candle_files']:
    raise ValueError('Manifest must contain both tick and candle partitions.')

missing_files = [item['name'] for key in ('tick_files', 'candle_files') for item in manifest[key] if not (DATASET / item['name']).is_file()]
if missing_files:
    preview = ', '.join(missing_files[:5])
    raise FileNotFoundError(f'Archive partitions are missing ({len(missing_files)} total): {preview}')

db = duckdb.connect()
CANDLE_GLOB = str(DATASET / 'candles-*.parquet')
available_days = [date.fromisoformat(item['day']) for item in manifest['tick_files']]
symbols = [item['symbol'] if isinstance(item, dict) else item for item in manifest['symbols']]
default_symbol = 'AAPL' if 'AAPL' in symbols else symbols[0]
```

## Archive overview

```python
archive_bytes = sum(item['bytes'] for key in ('tick_files', 'candle_files') for item in manifest[key])
overview = pd.DataFrame({
    'value': [manifest['dataset_id'], len(symbols), available_days[0].isoformat(), available_days[-1].isoformat(),
              f"{manifest['tick_count']:,}", f"{manifest['candle_count']:,}",
              len(manifest['tick_files']), len(manifest['candle_files']), f'{archive_bytes / 1024**3:.2f} GiB']
}, index=['Dataset', 'Symbols', 'First day', 'Last day', 'Archived ticks', 'Candles',
          'Tick partitions', 'Candle partitions', 'Manifest archive size'])
display(overview)
display(HTML(f"<b>Symbols:</b> {', '.join(symbols)}<br><b>Archive:</b> {DATASET}"))
```

## Query helpers

DuckDB applies symbol and timestamp filters while scanning Parquet. The browser receives only the selected symbol's aggregated candle rows or one day of ticks.

```python
RANGE_SETTINGS = {
    'Day': ('1 minute', 'day'),
    'Month': ('15 minutes', 'month'),
    '3 months': ('1 hour', 'three_months'),
    'Year': ('1 day', 'year'),
}

def chicago_epoch(value: datetime) -> int:
    return int(value.replace(tzinfo=CHICAGO).timestamp())

def range_bounds(anchor: date, range_name: str) -> tuple[datetime, datetime]:
    mode = RANGE_SETTINGS[range_name][1]
    if mode == 'day':
        start = datetime.combine(anchor, datetime.min.time())
        end = start + timedelta(days=1)
    elif mode == 'month':
        start = datetime(anchor.year, anchor.month, 1)
        end = (pd.Timestamp(start) + pd.offsets.MonthBegin(1)).to_pydatetime()
    elif mode == 'three_months':
        start = datetime(anchor.year, anchor.month, 1)
        end = (pd.Timestamp(start) + pd.DateOffset(months=3)).to_pydatetime()
    else:
        start, end = datetime(anchor.year, 1, 1), datetime(anchor.year + 1, 1, 1)
    return start, end

def candle_range(symbol: str, anchor: date, range_name: str) -> tuple[pd.DataFrame, datetime, datetime]:
    bucket, _ = RANGE_SETTINGS[range_name]
    start, end = range_bounds(anchor, range_name)
    query = f'''
        SELECT time_bucket(INTERVAL '{bucket}', to_timestamp(t)) AS timestamp,
               arg_min(open, t)::DOUBLE AS open, max(high)::DOUBLE AS high,
               min(low)::DOUBLE AS low, arg_max(close, t)::DOUBLE AS close,
               sum(volume)::BIGINT AS volume, sum(trade_count)::BIGINT AS trade_count
        FROM read_parquet(?)
        WHERE symbol = ? AND t >= ? AND t < ?
        GROUP BY 1 ORDER BY 1
    '''
    frame = db.execute(query, [CANDLE_GLOB, symbol, chicago_epoch(start), chicago_epoch(end)]).fetch_df()
    if frame.empty:
        raise ValueError(f'No candle data for {symbol} in the selected {range_name.lower()} range.')
    return frame, start, end

def add_condition_regions(fig: go.Figure, start: datetime, end: datetime) -> None:
    colors = {'normal': '#94a3b8', 'uptrend': '#22c55e', 'downtrend': '#ef4444',
              'sideways': '#3b82f6', 'consolidation': '#a855f7', 'volatility_spike': '#f59e0b'}
    config = manifest.get('config', {})
    for event in config.get('schedule', []) + config.get('overlays', []):
        event_start, event_end = pd.Timestamp(event['start']).to_pydatetime(), pd.Timestamp(event['end']).to_pydatetime()
        if event_end > start and event_start < end:
            fig.add_vrect(x0=max(event_start, start), x1=min(event_end, end), fillcolor=colors.get(event['condition'], '#64748b'),
                          opacity=0.08, line_width=0, annotation_text=event['condition'], annotation_position='top left')
```

## Interactive candle range

```python
symbol_control = widgets.Dropdown(options=symbols, value=default_symbol, description='Symbol')
range_control = widgets.ToggleButtons(options=list(RANGE_SETTINGS), value='Year', description='Range')
date_control = widgets.DatePicker(value=available_days[0], description='Anchor date')

def show_candles(symbol: str, range_name: str, anchor_date: date) -> None:
    if anchor_date is None:
        print('Choose an anchor date.')
        return
    try:
        frame, start, end = candle_range(symbol, anchor_date, range_name)
    except (ValueError, duckdb.Error) as error:
        print(f'Unable to render candle view: {error}')
        return
    fig = make_subplots(rows=2, cols=1, shared_xaxes=True, vertical_spacing=0.04, row_heights=[0.72, 0.28])
    fig.add_trace(go.Candlestick(x=frame['timestamp'], open=frame['open'], high=frame['high'],
                                 low=frame['low'], close=frame['close'], name=symbol), row=1, col=1)
    fig.add_trace(go.Bar(x=frame['timestamp'], y=frame['volume'], name='Volume', marker_color='#64748b'), row=2, col=1)
    add_condition_regions(fig, start, end)
    fig.update_layout(title=f'{symbol}: {range_name} ({RANGE_SETTINGS[range_name][0]} resolution)',
                      height=700, template='plotly_white', xaxis_rangeslider_visible=False, hovermode='x unified')
    fig.update_yaxes(title_text='Price', row=1, col=1)
    fig.update_yaxes(title_text='Volume', row=2, col=1)
    fig.show()
    display(HTML(f'<b>{len(frame):,}</b> aggregated points from <b>{start.date()}</b> through <b>{(end - timedelta(days=1)).date()}</b>.'))

candle_output = widgets.interactive_output(show_candles,
    {'symbol': symbol_control, 'range_name': range_control, 'anchor_date': date_control})
display(widgets.VBox([widgets.HBox([symbol_control, range_control, date_control]), candle_output]))
```

## Full-year comparison and daily volatility

```python
daily = db.execute('''
    SELECT symbol, time_bucket(INTERVAL '1 day', to_timestamp(t)) AS timestamp,
           arg_max(close, t)::DOUBLE AS close, sum(volume)::BIGINT AS volume
    FROM read_parquet(?) GROUP BY symbol, 2 ORDER BY symbol, 2
''', [CANDLE_GLOB]).fetch_df()
daily['normalized'] = daily.groupby('symbol')['close'].transform(lambda values: values / values.iloc[0] * 100)
performance = go.Figure()
for symbol, frame in daily.groupby('symbol'):
    performance.add_trace(go.Scatter(x=frame['timestamp'], y=frame['normalized'], name=symbol, mode='lines'))
performance.update_layout(title='2026 normalized performance (first close = 100)', yaxis_title='Normalized close',
                          height=550, template='plotly_white', hovermode='x unified')
add_condition_regions(performance, datetime(2026, 1, 1), datetime(2027, 1, 1))
performance.show()

selected_daily = daily[daily['symbol'] == default_symbol].copy()
selected_daily['daily_return_pct'] = selected_daily['close'].pct_change() * 100
selected_daily['rolling_volatility_pct'] = selected_daily['daily_return_pct'].rolling(20, min_periods=5).std()
risk = make_subplots(rows=2, cols=1, shared_xaxes=True, vertical_spacing=0.08)
risk.add_trace(go.Bar(x=selected_daily['timestamp'], y=selected_daily['daily_return_pct'], name='Daily return'), row=1, col=1)
risk.add_trace(go.Scatter(x=selected_daily['timestamp'], y=selected_daily['rolling_volatility_pct'],
                          name='20-session volatility', line={'color': '#f59e0b'}), row=2, col=1)
risk.update_layout(title=f'{default_symbol}: daily returns and rolling volatility', height=600, template='plotly_white')
risk.update_yaxes(title_text='Return (%)', row=1, col=1)
risk.update_yaxes(title_text='Std. dev. (%)', row=2, col=1)
risk.show()
```

## One-second tick explorer

Only the selected day's tick file is opened. For one symbol this is 23,400 rows covering 08:30:00 through 14:59:59 America/Chicago.

```python
tick_symbol_control = widgets.Dropdown(options=symbols, value=default_symbol, description='Symbol')
tick_day_control = widgets.SelectionSlider(options=[(day.isoformat(), day) for day in available_days],
                                           value=available_days[0], description='Day', continuous_update=False,
                                           layout=widgets.Layout(width='700px'))
tick_file_by_day = {date.fromisoformat(item['day']): DATASET / item['name'] for item in manifest['tick_files']}

def load_day_ticks(symbol: str, day: date) -> pd.DataFrame:
    path = tick_file_by_day.get(day)
    if path is None or not path.is_file():
        raise FileNotFoundError(f'Tick partition for {day} is missing. Regenerate or restore the validated archive.')
    return db.execute('''
        SELECT to_timestamp(t) AS timestamp, price::DOUBLE AS price, bid::DOUBLE AS bid, ask::DOUBLE AS ask,
               bid_size, ask_size, trade_volume, sequence_number
        FROM read_parquet(?) WHERE symbol = ? ORDER BY t
    ''', [str(path), symbol]).fetch_df()

def show_ticks(symbol: str, day: date) -> None:
    try:
        frame = load_day_ticks(symbol, day)
    except (FileNotFoundError, duckdb.Error) as error:
        print(f'Unable to render tick view: {error}')
        return
    if frame.empty:
        print(f'No ticks found for {symbol} on {day}.')
        return
    fig = make_subplots(rows=2, cols=1, shared_xaxes=True, vertical_spacing=0.05, row_heights=[0.72, 0.28])
    fig.add_trace(go.Scattergl(x=frame['timestamp'], y=frame['bid'], name='Bid', line={'width': 1, 'color': '#3b82f6'}), row=1, col=1)
    fig.add_trace(go.Scattergl(x=frame['timestamp'], y=frame['ask'], name='Ask', line={'width': 1, 'color': '#ef4444'}), row=1, col=1)
    fig.add_trace(go.Scattergl(x=frame['timestamp'], y=frame['price'], name='Trade', line={'width': 1.2, 'color': '#111827'}), row=1, col=1)
    fig.add_trace(go.Bar(x=frame['timestamp'], y=frame['trade_volume'], name='Trade volume', marker_color='#64748b'), row=2, col=1)
    fig.update_layout(title=f'{symbol}: one-second ticks on {day}', height=700, template='plotly_white', hovermode='x unified')
    fig.update_yaxes(title_text='Price', row=1, col=1)
    fig.update_yaxes(title_text='Volume', row=2, col=1)
    fig.show()
    first, last = frame['timestamp'].iloc[0], frame['timestamp'].iloc[-1]
    display(HTML(f'<b>{len(frame):,} ticks</b> from {first} through {last}; sequences unique: <b>{frame.sequence_number.is_unique}</b>.'))
    display(frame.head(20))

tick_output = widgets.interactive_output(show_ticks, {'symbol': tick_symbol_control, 'day': tick_day_control})
display(widgets.VBox([widgets.HBox([tick_symbol_control, tick_day_control]), tick_output]))
```

## Interpretation

The annual and range views intentionally use candles because charting tens of millions of tick points is neither readable nor memory-efficient. The one-second records remain intact in daily Parquet partitions and can support a future replay service or focused analysis like the selected-day explorer above.
